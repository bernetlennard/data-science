```python
import pandas as pd

df = pd.read_csv(
    'shoporders.csv',
    sep=',',
    encoding='utf-8',
    decimal='.',
    parse_dates=['order_date']
)

df
```

```python
df['shipping_country'] = df['shipping_country'].str.strip().str.title()

df['shipping_country'] = df['shipping_country'].replace({
    'Switerland': 'Schweiz',
    'Switzerlaand': 'Schweiz',
    'Deuschland': 'Deutschland',
    'Östereich': 'Österreich'
})

print("Kontrolle der Länder nach Bereinigung:")
print(df['shipping_country'].value_counts())
```

```python
df['shipping_country'] = df['shipping_country'].fillna('Unbekannt')
df['customer_name'] = df['customer_name'].fillna('Unbekannt')
df['price'] = df['price'].fillna(df['price'].mean())

print(df.isna().sum())
```

```python
print("Anzahl Zeilen vor Dubletten-Löschung:", len(df))
print("Gefundene Dubletten:", df.duplicated().sum())
df = df.drop_duplicates(keep='first')
print("Anzahl Zeilen nach Bereinigung:", len(df))
```

```python
# Frage 1: Wie viele Bestellungen kommen aus der Schweiz?
schweiz_bestellungen = df['shipping_country'].value_counts().get('Schweiz', 0)
print("1. Bestellungen aus der Schweiz:", schweiz_bestellungen)

# Frage 2: Filtern Sie alle Bestellungen heraus, bei denen mehr als 3 Artikel bestellt wurden.
mehr_als_3 = df[df['quantity'] > 3]
print("2. Anzahl Bestellungen mit > 3 Artikeln:", len(mehr_als_3))

# Frage 3: In welchem Monat wurden die meisten Bestellungen aufgegeben?
haeufigster_monat = df['order_date'].dt.month_name().value_counts().idxmax()
print("3. Monat mit den meisten Bestellungen:", haeufigster_monat)
```
