```python
import pandas as pd

df = pd.read_csv(
    'shoporders.csv',
    sep=',',
    encoding='utf-8',
    decimal='.',
    parse_dates=['order_date']
)

df
```

```python
# Aufgabe 1, Spalte payment_method in One-Hot-Encoding umwandeln
df = pd.get_dummies(df, columns=['payment_method'], prefix='pay')
df
```

```python
# Aufgabe 2, Spalte order_date in weekday, month und is_weekend umwandeln
df['order_date'] = pd.to_datetime(df['order_date'])
df['weekday'] = df['order_date'].dt.dayofweek + 1
df['month'] = df['order_date'].dt.month
df['is_weekend'] = df['weekday'].isin([6, 7])
df
```

```python
# Aufgabe 3, Verhältnis Feature
df['price_per_unit'] = df['price'] / df['quantity']
df
```

```python
# Aufgabe 4, Numerische Spalte skalieren
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df['price_scaled'] = scaler.fit_transform(df[['price']])
df
```

```python
# Aufgabe 5, Werte Binnen
labels = ['<= 10', '11-50', '50-100', '100-200', '200-500', '> 500']
bins = [0, 11, 51, 101, 201, 501, 100000000]
df['price_binned'] = pd.cut(df['price'], bins=bins, labels=labels)
df
```

```python
# Aufgabe 6,Gruppenbasierte Aggregation
df['customer_avg_order'] = (df.groupby('customer_name')['price'].transform('mean'))
df
```

```python
# Aufgabe 7, Log Transformation
import numpy as np
df['price_log'] = np.log1p(df['price'])
df
```
